```python
import pandas 
import numpy
import re # this module provides pattern matching operations among list of strings given
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB  # word counts for text classification
from sklearn.metrics import accuracy_score
```

```python
data = pandas.read_csv("twitter_validation.csv",header = None)
df = pandas.DataFrame(data)
df
```

```python
df.columns = ["ID","PLATFORM","NATURE","COMMENT"]
df.head()
```

```python
df["COMMENT"] = df["COMMENT"].str.lower()
df["COMMENT"]
```

```python
# re.sub(replacement , "", text)

def replace_url(text):

    text = re.sub(r'http\S+',"",text) # replace all urls starting with http or https with empty string
    # \S+  = Matches any non-whitespace character
    text = re.sub(r'[^a-z\s]',"",text)
    # \s   = Matches a single whitespace character 
    return text
df["COMMENT"] = df["COMMENT"].apply(replace_url)

df["COMMENT"]
```

```python
# import word_tokenize

df["Tokens"] = df["COMMENT"].apply(word_tokenize)
df["Tokens"]

```

```python
# remove stop words

stop_words = set(stopwords.words("english"))
df["Tokens"] = df["Tokens"].apply(lambda a : [i for i in a if i not in stop_words])
df["Tokens"]
```

```python
# lemmatization

lemmatizer = WordNetLemmatizer() # method called lematize used for stemming each tokens in each row
# Lemmatize as verbs
df["Tokens_verb"] = df["Tokens"].apply(lambda a: [lemmatizer.lemmatize(i, pos="v") for i in a])

# Lemmatize as nouns
df["Tokens_noun"] = df["Tokens"].apply(lambda a: [lemmatizer.lemmatize(i, pos="n") for i in a])

# Lemmatize as adjectives
df["Tokens_adj"] = df["Tokens"].apply(lambda a: [lemmatizer.lemmatize(i, pos="a") for i in a])

# Lemmatize as adverbs
df["Tokens_adv"] = df["Tokens"].apply(lambda a: [lemmatizer.lemmatize(i, pos="r") for i in a])

df["Tokens"]
```

```python
# stemming 

obj = PorterStemmer()
df["Tokens"] = df["Tokens"].apply(lambda a : [obj.stem(i) for i in a])
df["Tokens"]
```

```python
# join all the tokens in each list back to a sentence

df['clean_text'] = df['Tokens'].apply(lambda x:"".join(x))
df['clean_text']
```

```python
tfid = TfidfVectorizer()
X = tfid.fit_transform(df['clean_text'])
```

```python
le = LabelEncoder()
y = le.fit_transform(df['NATURE'])
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)

# Navie bayes model
model = MultinomialNB()
model.fit(X_train,y_train)

y_pred = model.predict(X_test)
y_pred
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

```python
# custom prediction
# =================

def custom(text):

    text = text.lower()
    text = replace_url(text)
    text = " ".join([
        obj.stem(i) for i in word_tokenize(text) if i not in stop_words])

    result = tfid.transform([text])
    pred = model.predict(result)

    return le.inverse_transform(pred)

print(custom("Python will give you error if you skip the indentation"))
```
