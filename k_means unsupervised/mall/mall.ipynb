```python
import pandas
import numpy
```

```python
data = pandas.read_csv("Mall_Customers.csv")
df = pandas.DataFrame(data)
df
```

```python
df = df.drop(columns=["CustomerID"],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
df
```

```python
df.dtypes
```

```python
x = df[["Annual Income (k$)","Spending Score (1-100)"]]
x
```

```python
# from sklearn.cluster import KMeans

# wcss = []

# for i in range(1,11):
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2,init = 'k-means++')

y_pred = kmeans.fit_predict(x)
y_pred
```

```python
from matplotlib import pyplot

pyplot.scatter(x.iloc[:,0],x.iloc[:,1],c=y_pred)
pyplot.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1])
pyplot.xlabel("Annual Income (k$)")     
pyplot.ylabel("Spending Score (1-100)")
pyplot.title("K means clustering")
pyplot.show()
```
