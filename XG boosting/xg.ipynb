# XG Boosting 
===============

# Extreme gradient boosting

>> one tree bulit and when the error occurred , the next tree is formed by focusing on the error , 
   like that the last tree is formed with zero error. 

>> underfitting , overfitting , error


```python
import pandas
from xgboost import XGBClassifier 

data = {
    "hours" : [1,2,4,5],
    "result": [0,1,1,1]        # 0 >> fail , 1 >> pass
}
```

```python
df = pandas.DataFrame(data)
df
```

```python
x = df[["hours"]]
y = df[["result"]]
```

```python
model = XGBClassifier()
model.fit(x,y)
y_pred = model.predict(x)
y_pred
```

```python
y
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y,y_pred))
```
