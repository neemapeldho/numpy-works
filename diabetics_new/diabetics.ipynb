```python
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
x
```

```python
# imblearn  >>>this package used to balance the rows and columns
# >> tool for imbalanced dataset in machine learing


# numpy >> vectorised calculation (numerical)
# pandas >> analyse , preprocessing
# sklearn >> machine learning
# nltk  >> nlp (natural language processing)
# matplotlib >> creating static, animated data visualization
# seaborn >> statistical graphics and data visualization
```

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE(random_state=42)
x_new,y_new = sm.fit_resample(x,y)

pandas.DataFrame(y_new).value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x_new,y_new,test_size=0.3)
```

```python
from sklearn.preprocessing import MinMaxScaler
obj = MinMaxScaler()
obj.fit_transform(x_train)
obj.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=7)
naive = GaussianNB()
objects = [knn,naive]

for i in objects:

    i.fit(x_train,y_train)
    y_pred = i.predict(x_test)

    print(f" Accuracy score if {i} is {accuracy_score(y_test,y_pred)}")
    print("===============================================================")
```
